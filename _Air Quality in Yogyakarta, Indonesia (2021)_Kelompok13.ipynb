```python
print("""STATISTIKA DESKRIPTIF VARIABEL NUMERIK""")
import pandas as pd
import numpy as np

kolom_numerik = ["PM2.5", "PM10", "CO"]

def statistik_deskriptif(df, kolom_list):
    ringkasan = {}
    for col in kolom_list:
        data = df[col].dropna()
        q1 = data.quantile(0.25)
        q3 = data.quantile(0.75)
        iqr = q3 - q1
        
        mode_val = data.mode()[0] if not data.mode().empty else np.nan
        
        ringkasan[col] = {
            "Mean (Rata-rata)": data.mean(),
            "Median": data.median(),
            "Modus": mode_val,
            "Minimum": data.min(),
            "Maksimum": data.max(),
            "Jangkauan (Range)": data.max() - data.min(),
            "Varians": data.var(),
            "Simpangan Baku (Std Dev)": data.std(),
            "Q1 (Kuartil 1)": q1,
            "Q3 (Kuartil 3)": q3,
            "IQR": iqr
        }
    return pd.DataFrame(ringkasan)

tabel_ringkasan = statistik_deskriptif(df, kolom_numerik)
display(tabel_ringkasan.round(2))
```

```python
print("""HITUNGAN VARIANS ATAU SIMPANGAN BAKU""")
import numpy as np
import math


kolom_list = ["PM2.5", "PM10", "CO"]

for col in kolom_list:
    print(f"\nPerhitungan Manual Varians 10 Baris Pertama ({col}) ")
    sample_10 = df[col].dropna().head(10).tolist()

    n = len(sample_10)
    mean_manual = sum(sample_10) / n
    deviasi_kuadrat = [(x - mean_manual)**2 for x in sample_10]
    var_manual = sum(deviasi_kuadrat) / (n - 1)
    std_manual = math.sqrt(var_manual) if 'math' in globals() else np.sqrt(var_manual)

    print(f"Data 10 baris pertama : {sample_10}")
    print(f"Rata-rata (mean)     : {mean_manual:.2f}")
    print(f"Varians Manual (n-1)  : {var_manual:.2f}")
    print(f"Varians Pandas       : {df[col].dropna().head(10).var():.2f}")
    print(f"Std Deviasi Manual   : {std_manual:.2f}")
    print(f"Std Deviasi Pandas   : {df[col].dropna().head(10).std():.2f}")
```

1.1.Suhu & Kelembapan: Kelembapan tinggi sering kali memicu pengendapan polutan di permukaan bumi atau mempercepat reaksi pembentukan partikel sekunder. Sebaliknya, suhu udara tinggi dapat memicu konveksi termal yang membantu dispersi polutan.

2.3 Efek Akhir Pekan (Weekend Effect): Di beberapa kota, rata-rata harian polutan seperti $\text{CO}$ dan $\text{PM}_{2.5}$ pada hari Minggu sedikit lebih rendah dibanding hari kerja karena berkurangnya aktivitas industri dan lalu lintas penglaju (commuter).
